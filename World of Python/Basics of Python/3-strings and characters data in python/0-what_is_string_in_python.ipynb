### What is String in python?

- String is a fundamental data type in python. It means that it can’t be broken up into smaller values of a different type. Those are actually called compound data types and are commonly known as data structures. Future Python Basics courses will dig into data structures

- The string data type has a special abbreviated name in Python, str. To see it, you can use the built-in type() function to show the data type of a given value.
- it can be created with single quote, double quote or tripple quote



```python
type('hello')
```

```python
phrase = 'Hello world!'
print(phrase)
```

```python
# you can use single quote and double quote togather inside string
str1 = "Hello Mr James, what's up??"
print(str1)
```

```python
str2 = 'what has happened during summer in the year "2024"?'
print(str2)
```

```python
str3 = 'what has happened during summer in the year '2024''?
print(str3)
```

```python
# to fix above syntex error, you can either use double quotes or use escape character inside string
str4 = 'what has happened during summer in the year \'2024\'?'
print(str4)
```

```python
## to check the length of a string you can use len() function
len(str4)
```

```python
## multiline strings can be created using two ways. 1> by escape '\' character or using tripple quotes.
#if you use escape character, it will print the entire multi line string in a single line

str5 = "This is a multiline string example.\
    As you can see it is printing\
        entrie multiline string into one line"
        
print(str5)
print(len(str5))
```

```python
##if you use tripple quotes, it will preserve the spaces and print the multiline string as it is. 
str6 = """This is a multiline string example.
    As you can see it is printing
        entrie multiline string into one line"""
        
print(str6)
print(len(str6))
```
