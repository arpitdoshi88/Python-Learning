## An operand can be either a literal value or a variable that references an object:

| Operator   | Example | Meaning                                       | Result                                                                                                                      |
|------------|---------|-----------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------|
| + (unary)  | +a      | Unary Positive                                | a, In other words, it doesn’t really do anything. It mostly exists for the sake of completeness, to complement Unary Negation.                                                                                                                          |
| + (binary) | a + b   | Addition                                      | Sum of a and b                                                                                                              |
| - (unary)  | -a      | Unary Negation                                | Value equal to a but opposite in sign                                                                                       |
| - (binary) | a - b   | Subtraction                                   | b subtracted from a                                                                                                         |
| *          | a * b   | Multiplication                                | Product of a and b                                                                                                          |
| /          | a / b   | Division                                      | Quotient when a is divided by b. The result always has type float.                                                                                               |
| %          | a % b   | Modulo                                        | Remainder when a is divided by b                                                                                            |
| //         | a // b  | Floor Division (also called Integer Division) | Quotient when a is divided by b, rounded to the next smallest whole number                                                  |
| **         | a ** b  | Exponentiation                                | a raised to the power of b                                                                                                  |

```python
a = 9
b = 2

print('+(unary)',+a)
print('+(binary)',a+b)
print('-(unary)',-a)
print('-(binary)',a-b)
print('multiplication',a*b)
print('division',a/b)
print('modulo',a%b)
print('floor division',a//b)
print('exponentiation',a**2)
```

#### The result of standard division (/) is always a float, even if the dividend is evenly divisible by the divisor:

```python
print(8/4)
print(type(8/4))
```

#### When the result of floor division (//) is positive, it is as though the fractional portion is truncated off, leaving only the integer portion. 
#### When the result is negative, the result is rounded down to the next smallest (greater negative) integer:




```python
print(10/4)
print(10//4)
print(10//-4)
print(-10//4)
print(-10//-4)
```

```python
# promotion of int to float
a = 5   
print(a+2.5)
print(type(a+2.5))

# auto conversion of int to float in mixed arithmetic operations
a = 4
b = 3.5
print(a + b)      # int + float = float
print(a - b)      # int - float = float
print(a * b)      # int * float = float
print(a / b)      # int / float = float
print(a // b)     # int // float = float
print(a ** b)     # int ** float = float        
```

```python
# integer division always returns floor value. observe the difference below for positive and negative numbers
print(7/3)
print(int(7/3)) # truncates towards zero
print(7//3)
print(-7/3)
print(int(-7/3)) # truncates towards zero
print(-7//3) # floor value
```

```python
import math

math.floor(-7/-3)
```

```python
# impact of negative numbers on modulo operator
print(7 % 3)    # 1
print(-7 % 3)   # 2 # result is always positive when divisor is positive. formula: dividend - (divisor * floor(dividend/divisor)) = -7 - (3 * -3) = 2
print(7 % -3)   # -2 # result is always negative when divisor is negative. formula: dividend - (divisor * floor(dividend/divisor)) = 7 - (-3 * -3) = -2
print(-7 % -3)  # -1 # result is always negative when divisor is negative. formula: dividend - (divisor * floor(dividend/divisor)) = -7 - (-3 * 2) = -1   
```

## Summary & Exercises

**Key takeaways:**

- `/` always performs true division and returns a `float`.
- `//` performs floor division: it returns the largest integer less than or equal to the exact quotient (i.e., it rounds down). With negative numbers this means the result may be *more negative* than simple truncation toward zero.
- `%` returns the remainder such that: `dividend == divisor * (dividend // divisor) + (dividend % divisor)`. This definition explains the sign behavior of `%` when operands are negative.
