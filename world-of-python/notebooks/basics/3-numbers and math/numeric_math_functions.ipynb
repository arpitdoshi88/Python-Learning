```python
print(round(1.3)) # rounds down
print(round(1.7)) # rounds up
print(round(1.5)) # rounds to earest even number
print(round(2.5)) # rounds to nearest even number
print(round(3.5)) # rounds to nearest even number
print(round(-1.5)) # rounds to nearest even number
print(round(-2.5)) # rounds to nearest even number
print(round(-3.5)) # rounds to nearest even number
```

```python
print(pow(2, 3))  # 8 same as exponentiation operator
print(2 ** 3)   # 8
print(pow(9, 0.5))  # 3.0 same as square root
print(9 ** 0.5)  # 3.0 #same as square root


import math
math.sqrt(9)  # 4.0
```
