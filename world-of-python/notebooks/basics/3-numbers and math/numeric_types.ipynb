#### Python has three numeric types built into its syntax, which means that you can start using them right away when you run IDLE. 

- The three numeric types native to Python are integers, or whole numbers, such as -3, 0, or 42.Python calls them int for short. Then there are floating-point numbers, which might have a fractional part, such as -2.72 or 3.14, for example. These are known as floats, and you’ll learn why they are called like that. Finally, there are complex numbers composed of the real and imaginary parts.

- All three numeric types built into Python are signed, which means they can store both positive or negative values along with the neutral zero. Additionally, integers don’t have a maximum or minimum value, which is only limited by the available memory on your computer. In contrast, floating-point numbers and their complex number cousins do have a fixed range of values as well as precision. Nevertheless, they’re big enough to accommodate most real-life use cases.

### express a number in different bases

```python
# binary base 2
# use prefix 0b or 0B for binary literals
b = 0b1010  # 1*2^3 +0*2^2 +1*2^1 +0*2^0 = 8 +0 +2 +0 = 10
print(b, type(b), sep=" -> ")
```

```python
# octal base 8
# use prefix 0o or 0O for octal literals
o = 0o52  # 5*8^1 + 2*8^0 = 40 +2 = 42
print(o, type(o), sep=" -> ")
```

```python
# hexadecimal. base 16
# use prefix 0x or 0X for hexadecimal literals
h = 0x2A  # 2*16^1 + 10*16^0 = 32 + 10 = 42
print(h, type(h), sep=" -> ")
```

```python
## use of int() to convert other types to int
## int() function assumes that you will supply a string consisting of decimal digits 0 to 9.

n = int("12345")
print(n, type(n), sep=" -> ")
```

```python
n1 = int(12.99)  # truncates the decimal part
print(n1, type(n1))
```

```python
n2 = int(True)  # True is converted to 1 and False to 0
print(n2, type(n2))

n3 = int(False)
print(n3, type(n3))
```

```python
n4 = int("1010", 2)  # binary to decimal
print(n4, type(n4))

n5 = int("52", 8)  # octal to decimal
print(n5, type(n5))

n6 = int("2A", 16)  # hexadecimal to decimal
print(n6, type(n6))
```

```python
n7 = int("ABC")  # int() assumes string consists of decimal digits only 0-9
print(n7, type(n7))
```

```python
# call int() without any arguments will return 0 always
n8 = int()
print(n8, type(n8))
```

```python
# range of int type depends on the system architecture
import sys

print(sys.maxsize)  # maximum size of int on this system
print(-sys.maxsize - 1)  # minimum size of int on this system
```

```python
i = sys.maxsize + 100
print(
    i, type(i)
)  # Python int can grow beyond sys.maxsize. hence no overflow error. The reasion is that Python int is of arbitrary precision.So it can grow as big as the memory allows.
```

### Floating point numbers

```python
# floating point numbers have whole number part and fractional part separated by a decimal point.
f = 3.1415
print(f, type(f), sep=" -> ")
```

```python
f1 = 0.25  # you don't need to put 0 before decimal point
print(f1, type(f1), sep=" -> ")
print(f1.real, f1.imag)  # real and imaginary parts of float number

f1 = -0.75  # negative float number
print(f1, type(f1), sep=" -> ")
```

```python
f3 = 1.25_38323_234  # underscores for readability of big float numbers
print(f3, type(f3))
```

```python
# floating point numbers can also be represented in scientific notation using e or E to indicate the power of 10 by which the number is multiplied.
f4 = 1.5e3  # 1.5 * 10^3 = 1.5 * 1000 = 1500.0
print(f4, type(f4), sep=" -> ")

f5 = 2.5e-4  # 2.5 * 10^-4 = 2.5 * 0.0001 = 0.00025
print(f5, type(f5), sep=" -> ")
```

```python
# create a float from int and string using float() function
f6 = float(10)  # int to float conversion
print(f6, type(f6), sep=" -> ")

f7 = float("3.1415")  # string to float conversion
print(f7, type(f7), sep=" -> ")

# float() called without any arguments returns 0.0
f8 = float()
print(f8, type(f8), sep=" -> ")
```

```python
# You’ll also want to use the float() function to define a few special values which don’t have literal representations in the floating-point number data type.
f_nan = float("nan")  # Not a Number
print(f_nan, type(f_nan))
f_inf = float("inf")  # Positive Infinity
print(f_inf, type(f_inf))
f_ninf = float("-inf")  # Negative Infinity
print(f_ninf, type(f_ninf))
```

```python
##nan follows IEEE 754 standard for floating point arithmetic. In this standard, NaN is not considered equal to any value, including itself. Therefore, any comparison involving NaN will return False, except for the "is" operator which checks for identity.
print(f_nan == f_nan)  # False
print(f_nan is f_nan)  # True
```

```python
## maximum and minimum float values
import sys

print(sys.float_info.max)  # Maximum representable positive finite float
print(sys.float_info.min)  # Minimum representable positive normalized float
```

### complex types

```python
a = 1.7j + 308
print(a, type(a))  # are complex numbers
a1 = complex(2, 3)  # complex number with real part 2 and imaginary part 3
print(a1, type(a1))
a2 = complex(0, -1)  # complex number with real part 0 and imaginary part -1
print(a2, type(a2))
print(a2.real, a2.imag)  # real and imaginary parts of complex number
```

```python
import math

for x in [float("nan"), float("inf"), 1.0]:
    print(f"x: {x}, x == x: {x == x}, x is x: {x is x}, math.isnan(x): {math.isnan(x)}")
```

## Summary & Exercises

**Key takeaways:**

- `int` represents whole numbers and has arbitrary precision (limited by memory).
- `float` follows IEEE 754 semantics; it supports `NaN` and infinite values — comparisons involving `NaN` are always False (except identity checks).
- `complex` numbers store a real and imaginary part; use `.real` and `.imag` to access them.
