### String concatination
-  To combine, or concatenate, strings, you use the + (plus) operator. You’re adding one string to another, and that concatenates them.

```python
print('Hello' + ' ' + 'world!')
```

```python
ip = input('Enter your name')
str1 = "Hello " + ip
print(str1)
```

In Python, strings are ordered sequences of character data, and thus can be indexed in this way. Individual characters in a string can be accessed by specifying the string name followed by a number in square brackets ([]).

String is immutable object and indexing in Python String is zero-based: the first character in the string has index 0, the next has index 1, and so on. The index of the last character will be the length of the string minus one.

```python
s = "Adam and Eva"
print(len(s))
```

```python
print(s[0]) ##first character
print(s[len(s)-1]) #last character
```

```python
#Attempting to index beyond the end of the string results in an error: 
s[13]
```

```python
s[len(s)]
```

String indices can also be specified with negative numbers, in which case indexing occurs from the end of the string backward: -1 refers to the last character, -2 the second-to-last character

```python
s[-1] # to get the last char of a string
```

```python
s[-len(s)] # to print the first char of a string
```

```python
s[-13] 
```
