A pyhton namespace maps the names with the object. It works like a dictionary where keys are object names and values are the objects themselves. Namespaces organize variables and functions in a dedicated space, allowing you to use multiple instances of the same name without conflict, as long as they’re in different namespaces.

- Python namespaces serve as containers that map names to objects, allowing for organized access and management of variables, functions, classes, and objects in general.
- Namespace and scope differ in that a namespace maps names to objects, while a scope is the region of code where you can access a name.
- Python implements most namespaces using dictionaries, where each namespace’s lifecycle is tied to the execution context, such as global or local scopes.

a Python program can have up to four different types of namespaces:

- Built-In
- Global
- Local
- Enclosing or nonlocal

These namespaces have differing lifetimes. As Python executes a program, it creates namespaces as necessary and removes them when it no longer needs them. Typically, many namespaces will exist at any given time.

The Python global, local, and nonlocal namespaces are implemented as dictionaries. In contrast, the built-in namespace isn’t a dictionary but a module called builtins. This module acts as the container for the built-in namespace.

### Built in namespace

The built-in namespace contains the names of all of Python’s built-in objects. This namespace is available while the Python interpreter is running. So, you can access the names that live in this namespace at any time in your code without explicitly importing them.

You can list the objects in the built-in namespace with the dir() function using the __builtins__ object as an argument

```python
dir(__builtin__) #### Built in namespace
```

### Global namespace

The global namespace contains the names defined at the module level. Python creates a main global namespace when the main program’s body starts. This namespace remains in existence until the interpreter terminates.

Additionally, each module has its own global namespace. The interpreter creates a global namespace for any module that your program loads with the import statement. 

```python
number = 42 ### Global namespace
dir() #check the list of names defined in your current global scope
```

### Local Namespace

The Python interpreter creates a new and dedicated namespace whenever you call a function. This namespace is local to the function and exists only until the function returns:

```python
def my_function(number):
    result = number + 3
    print(dir())  ### Local Namespace. It shows names defined in the local scope of this function. number and result are local variables.
    return result

my_function(7) 
print(result) # This will give an error because result is not defined in the global scope.
```

### Enclosing and non-local Namespace

```python
global_var = "I am global!"

def outer_function():
    enclosing_var = "I am enclosing!"
    
    def inner_function():
        #nonlocal enclosing_var
        local_var = "I am local!"
        print(local_var)          # Local Namespace
        print(enclosing_var)      # Enclosing Namespace
        print(global_var)         # Global Namespace
    
    inner_function()    
    
outer_function()    
```

# Understanding the Scope of Names


The existence of multiple, distinct namespaces allows you to have several different instances of a particular name simultaneously while a Python program runs. As long as each instance is in a different namespace, they’re all maintained separately and won’t interfere with one another.

That raises a question. Suppose you refer to the name x in your code, and x exists in several namespaces. How does Python know which one you mean each time?

The answer lies in the concept of scope. The scope is the region of a program in which a name has meaning. The interpreter determines this at runtime based on where the name definition occurs and where the name is referenced in the code.

## The LEGB Rule for Searching Names
Returning to the question from the previous section, say that you’re referencing x from inside an inner function. In this situation, Python looks for x in the following order:

- Local: Python searches for x inside the inner function. If it doesn’t find x there, then it moves to the enclosing scope.
- Enclosing: Next, Python searches for x in the enclosing function’s scope. If it’s not found there, then it moves to the global scope.
- Global: Python searches for x in the global scope. If it still doesn’t find it, it moves to the built-in scope.
- Built-in: Finally, Python searches for x in the built-in scope. If it doesn’t find x there, then it raises a NameError exception.

```python
x = 'global x'

def outer():
    def inner():
        print(x) # This will print 'global x' because it looks in the global scope since there is no local or enclosing x.
    inner()
outer()
```

```python
x = 'global x'
def outer():
    x = 'enclosing x'
    def inner():
        print(x) # this will print 'enclosing x' because it finds x in the enclosing scope.
    inner()
    
outer()    
```

```python
x = 'global x'
def outer():
    x = 'enclosing x'
    def inner():
        x = 'local x'
        print(x) # This will print 'local x' because it finds x in the local scope first.
    inner()
    print(x) # This will print 'enclosing x' because it finds x in the enclosing scope.
outer()
print(x) # This will print 'global x' because it finds x in the global scope. 
```

```python
del x ### Deleting x from global scope

def outer():    
    def inner():
        print(x) # This will give an NameError because x is not defined in local or enclosing scope.
    inner()
outer()
```

### Managing Namespace Dictionaries

Python provides two built-in functions, globals() and locals(), that allow you to access the global and local namespace dictionaries. These functions give you direct access to the content of both namespaces, which are implemented as dictionaries.

### globals() function

The built-in globals() function returns a reference to the current global namespace dictionary. You can use it to access the objects in the global namespace

```python
globals()
```

```python
number = 42
globals() # This will show 'number' in the global namespace dictionary
```

```python
globals()['number'] is number  # This will return True, confirming both refer to the same object
```

```python
globals()['number'] = 100 # Modifying the global variable 'number' via globals() dictionary
print(number)  # This will print 100    
globals().pop('number') # Remove 'number' from the global namespace
```

```python
globals()['number'] # This will give an error because 'number' has been removed from the global namespace
```

### locals() function

```python
def my_function():
    ### Local Namespace
    local_var = 'I am local' # This is a local variable
    print(locals())  # This will show 'local_var' in the local namespace dictionary
    return local_var
my_function()
```

### The Difference Between globals() and locals()



There’s one slight difference between globals() and locals() that’s useful to know about. The globals() function returns a reference to the dictionary that contains the global namespace. Because of this, if you call globals() and save its return value, then you can add new variables and modify existing ones as you’ve already learned.

In contrast, locals() returns a dictionary that’s a copy of the dictionary that holds the current local namespace rather than a reference to it. Because of this subtle difference, additions to the return value of locals() won’t modify the actual namespace:

```python
def func():
    a = 10
    loc = locals()
    print(id(locals()), id(loc))
    print(f'{loc=}')
    b=20
    print(f'{loc=}')
    loc['c'] = 30  # Attempting to add a new variable 'c' to the local namespace
    print(f'{loc=}')  # This will show 'a', 'b', and 'c' in the local namespace dictionary
    print(f'{locals()=}')  # This will NOT show 'c' because locals() returns a copy of the local namespace
    return a + b

func()
```

it’s worth noting that locals() returns a shallow copy of the namespace, so any name that points to a mutable object lets you change that object in place. This change is reflected in the local namespace:

```python
def my_function():
    fruits = ['banana','mango'] # This is a local variable
    print(f'Before modification: {locals()}')
    loc = locals()
    loc['fruits'].append('peach')  # Modifying the mutable object in place
    print(f'After modification: {locals()}')
    print(f'{loc=}')
    
my_function()
```

### Modifying Variables From a Different Namespace

In a function, you can’t reassign a variable that points to an object defined on the outside and expect the change to affect that variable:

```python
x = 20

def func():
    x = 40
    print(x)  # This will print 40, the local variable
    
func()
print(x)  # This will print 20, the global variable
```

In contrast, a function can modify a mutable object from an outer scope:

```python
_list = [1, 2, 3]  # Using a different name to avoid shadowing the built-in 'list'
print(_list)  # This will print the user-defined list, not the built-in list type

def func():
    _list.append(4) # Modifying the global list which is mutable object
    print(_list)
    
func()
print(_list)  # This will print the modified list [1, 2, 3, 4]
```

### The global Statement

If you need to modify a value in the global scope from within a function, you can use global statement

```python
x = 20

def func():
    global x
    x = 40
    print(x)  # This will print 40, the local variable
    
func()
print(x) 
```

### The nonlocal statment

A similar situation occurs with nested functions when you need to modify variables in the enclosing scope. The nonlocal keyword addresses this by referring to variables in the nearest enclosing scope, which allows you to modify them within the nested function

```python
def f():
    enclosing_var = "I am enclosing!"
    
    def g():
        nonlocal enclosing_var # Use nonlocal to modify the enclosing variable
        enclosing_var = "Modified enclosing!"
        print(enclosing_var)  # This will print "Modified enclosing!"
    
    g()
    print(enclosing_var)  # This will also print "Modified enclosing!" because   it was modified in g()

f()
```
